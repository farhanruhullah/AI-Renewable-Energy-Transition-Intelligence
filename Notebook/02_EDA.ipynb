# Exploratory Data Analysis

```python
import pandas as pd
import plotly.express as px

df=pd.read_csv("../data/renewable_energy_transition_ENHANCED_FINAL.csv")
```

```python
trend=df.groupby("year")["renewables_share_elec"].mean().reset_index()
px.line(trend,x="year",y="renewables_share_elec",markers=True).show()
```

```python
df.corr(numeric_only=True)["co2_pc"].sort_values(ascending=False)
```
